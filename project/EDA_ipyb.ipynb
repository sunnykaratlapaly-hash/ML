```python
df = pd.read_csv("esports_matches.csv")

# Display first 5 rows
print(df.head())

# Dataset information
print(df.info())

# Statistical summary
print(df.describe())
```

```python
numerical_features = df.select_dtypes(
    include=['int64', 'float64']
).columns

categorical_features = df.select_dtypes(
    include=['object']
).columns

print("Numerical Features:", list(numerical_features))
print("Categorical Features:", list(categorical_features))
```

```python
print(df.isnull().sum())
```

```python
plt.figure(figsize=(12, 6))

sns.heatmap(
    df.isnull(),
    cbar=False,
    cmap="viridis"
)

plt.title("Missing Value Heatmap")
plt.show()
```

```python
numerical_cols = df.select_dtypes(
    include=[np.number]
).columns.tolist()

corr_matrix = df[numerical_cols].corr()

print("\n--- Correlation Matrix ---")
print(corr_matrix)
```

```python
plt.figure(figsize=(15, 12))

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    vmin=-1,
    vmax=1,
    square=True,
    linewidths=0.5
)

plt.title(
    "Correlation Heatmap of Esports Match Features",
    fontsize=14,
    fontweight="bold"
)

plt.tight_layout()
plt.show()
```

```python
features = [
    "TeamKills",
    "TeamDeaths",
    "TeamAssists",
    "GoldEarned",
    "GameDuration",
    "Objectives"
]

df[features].hist(
    bins=20,
    figsize=(15, 8)
)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="MatchOutcome"
)

plt.title("Distribution of Match Outcomes")
plt.xlabel("Match Outcome")
plt.ylabel("Number of Matches")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="MatchOutcome",
    y="TeamKills"
)

plt.title("Team Kills vs Match Outcome")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="MatchOutcome",
    y="GoldEarned"
)

plt.title("Gold Earned vs Match Outcome")

plt.show()
```

```python
numerical_columns = [
    "TeamKills",
    "TeamDeaths",
    "TeamAssists",
    "GoldEarned",
    "GameDuration"
]

plt.figure(figsize=(14, 8))

df[numerical_columns].boxplot()

plt.xticks(rotation=45)

plt.title("Boxplots of Esports Match Features")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    x="GameType",
    hue="MatchOutcome"
)

plt.title("Match Outcome by Game Type")

plt.xticks(rotation=45)

plt.show()
```
